```python
from modules import analisisEspacial
import pandas as pd
import importlib
import cmocean
importlib.reload(analisisEspacial)
```

```python
data2004 = pd.read_excel("./data/raw/oceanography/erfen_data/Pacífico_39_2004_Verano.xlsx", sheet_name="Hoja1").drop(index=0)
data2005 = pd.read_excel("./data/raw/oceanography/erfen_data/Pacífico_41_2005_Verano.xlsx", sheet_name="Hoja1").drop(index=0)
data2006 = pd.read_excel("./data/raw/oceanography/erfen_data/Pacífico_42_2006_Invierno.xlsx", sheet_name="Hoja1").drop(index=0)
```

```python
dataSup2004 = data2004[data2004['Profundidad[m]'] == 0]
dataSup2005 = data2005[data2005['Profundidad[m]'] == 0]
dataSup2006 = data2006[data2006['Profundidad[m]'] == 0]
```

```python
for data, year in zip([dataSup2004, dataSup2005, dataSup2006], [2004, 2005, 2006]):
     for algorithm in ['linear', 'cubic', 'nearest']:
            for variable, colour in zip(['Temperatura[°C]', 'Salinidad[UPS]'], [cmocean.cm.balance, cmocean.cm.haline]):
                print(f'Algoritmo: {algorithm} - Variable: {variable}')
                analisisEspacial.interpolacion(data, variable, year, color=colour, labelVar=None,
                                            tipo_interp=algorithm, rbf_function=False,
                                            validacion_cruzada=True, cv=-1)

```

```python
for data, year in zip([dataSup2004, dataSup2005, dataSup2006], [2004, 2005, 2006]):
     for algorithm in ['linear',  'gaussian', 'multiquadric', 'inverse', 'inverse_multiquadric']:
            for variable, colour in zip(['Temperatura[°C]', 'Salinidad[UPS]'], [cmocean.cm.balance, cmocean.cm.haline]):
                print(f'Algoritmo: {algorithm} - Variable: {variable}')
                analisisEspacial.interpolacion(data, variable, year, color=colour, labelVar=None,
                                            tipo_interp='rbf', rbf_function=algorithm,
                                            validacion_cruzada=True, cv=-1)
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Cargar imágenes
img1 = mpimg.imread("./outputs/mapas/Salinidad[UPS]_2004_linear_False_clipped.png")
img2 = mpimg.imread("./outputs/mapas/Salinidad[UPS]_2005_linear_False_clipped.png")
img3 = mpimg.imread("./outputs/mapas/Salinidad[UPS]_2006_linear_False_clipped.png")

# Tamaño físico de media carta en vertical (ancho x alto) en pulgadas
fig = plt.figure(figsize=(5.5, 8.5), dpi=300)
gs = fig.add_gridspec(3, 1, hspace=0.05)  # hspace pequeño para comprimir espacios

# Subplots
ax1 = fig.add_subplot(gs[0, 0])
ax1.imshow(img1)
ax1.axis("off")
ax1.set_title("A", loc="left", fontsize=8, fontweight="bold")

ax2 = fig.add_subplot(gs[1, 0])
ax2.imshow(img2)
ax2.axis("off")
ax2.set_title("B", loc="left", fontsize=8, fontweight="bold")

ax3 = fig.add_subplot(gs[2, 0])
ax3.imshow(img3)
ax3.axis("off")
ax3.set_title("C", loc="left", fontsize=8, fontweight="bold")

# Ajuste fino de márgenes
plt.tight_layout(pad=0.2)
plt.savefig("./outputs/mapas/salinidad_half_letter_vertical.png",
            dpi=300, bbox_inches="tight")
plt.show()
```
